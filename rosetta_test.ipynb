```python
import pyrosetta
from pyrosetta import init, pose_from_file
from pyrosetta.teaching import get_fa_scorefxn

# 1. PyRosetta 초기화
init("-mute all")

def analyze_structure(file_path, label):
    try:
        # 2. 구조 파일 불러오기
        pose = pose_from_file(file_path)

        # 3. 고해상도 물리력 스코어링 함수 불러오기
        scorefxn = get_fa_scorefxn()

        # 4. 전체 구조 에너지 계산
        total_score = scorefxn(pose)

        # 5. 세부 항목별 에너지 추출
        energies = pose.energies()
        fa_rep = energies.total_energies()[pyrosetta.rosetta.core.scoring.fa_rep]
        fa_elec = energies.total_energies()[pyrosetta.rosetta.core.scoring.fa_elec]

        print(f"[{label}]")
        print(f"  ▶ Total Score (총 에너지): {total_score:.2f} (낮을수록 안정)")
        print(f"  ▶ fa_rep (척력 페널티): {fa_rep:.2f} (비정상적으로 높으면 억지 결합)")
        print(f"  ▶ fa_elec (정전기력): {fa_elec:.2f}")
        print("-" * 55)

    except Exception as e:
        print(f"[{label}] 파일 읽기 오류: {e}")

print("=" * 55)
print("PyRosetta 스코어링 비교 테스트 시작...")
print("=" * 55)

analyze_structure("/content/wt.cif", "정상 서열 (WT)")
analyze_structure("/content/mt.cif", "돌연변이 서열 (Mutant)")
```
